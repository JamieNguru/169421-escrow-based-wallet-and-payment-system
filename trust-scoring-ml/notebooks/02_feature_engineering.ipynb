```python
import pandas as pd
import numpy as np
```

```python
transactions = pd.read_csv("../data/transactions_data.csv")

print("Transactions loaded:", transactions.shape)
```

```python
transactions = transactions[
    [
        "id",
        "date",
        "client_id",
        "card_id",
        "amount",
        "use_chip",
        "merchant_id",
        "mcc",
        "errors"
    ]
].copy()
```

```python
transactions["date"] = pd.to_datetime(transactions["date"])
```

```python
transactions.info()
```

```python
print(transactions["amount"].dtype)
print(transactions["amount"].head())
```

```python
transactions["amount"] = pd.to_numeric(
    transactions["amount"],
    errors="coerce"
)
```

```python
print("Amount data type:", transactions["amount"].dtype)
print("Missing amounts:", transactions["amount"].isna().sum())
```

```python
transactions["amount"] = (
    transactions["amount"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```
