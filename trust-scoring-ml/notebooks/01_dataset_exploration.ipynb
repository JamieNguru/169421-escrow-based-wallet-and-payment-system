# Dataset Exploration

```python
import pandas as pd
import json
```

```python
transactions = pd.read_csv("../data/transactions_data.csv")
users = pd.read_csv("../data/users_data.csv")
cards = pd.read_csv("../data/cards_data.csv")

print("Transactions:", transactions.shape)
print("Users:", users.shape)
print("Cards:", cards.shape)
```

```python
print("Transactions columns:")
print(transactions.columns.tolist())

print("\nUsers columns:")
print(users.columns.tolist())

print("\nCards columns:")
print(cards.columns.tolist())
```

```python
from pathlib import Path

mcc_path = Path("../data/mcc_codes.json")
fraud_path = Path("../data/train_fraud_labels.json")

print("MCC file size:",
      round(mcc_path.stat().st_size / (1024 * 1024), 2), "MB")

print("Fraud labels file size:",
      round(fraud_path.stat().st_size / (1024 * 1024), 2), "MB")
```

```python
with open(mcc_path, "r", encoding="utf-8") as file:
    print(file.read(1000))
```

```python
with open(fraud_path, "r", encoding="utf-8") as file:
    print(file.read(1000))
```

```python
transaction_counts = transactions["client_id"].value_counts()

print("Unique clients:", transactions["client_id"].nunique())
print("Clients with 1 transaction:", (transaction_counts == 1).sum())
print("Clients with 2 transactions:", (transaction_counts == 2).sum())
print("Clients with 3-4 transactions:", ((transaction_counts >= 3) & (transaction_counts <= 4)).sum())
print("Clients with 5+ transactions:", (transaction_counts >= 5).sum())
print("Clients with 10+ transactions:", (transaction_counts >= 10).sum())
```

```python
transaction_counts.describe()
```

```python
print("Total transactions:", len(transactions))
print("Unique clients:", transactions["client_id"].nunique())
print("Unique cards:", transactions["card_id"].nunique())
```

```python
transactions["date"] = pd.to_datetime(transactions["date"])

print("Earliest transaction:", transactions["date"].min())
print("Latest transaction:", transactions["date"].max())
```

```python
print(transactions["errors"].value_counts(dropna=False).head(20))
```

```python
print(
    "Transactions with errors:",
    transactions["errors"].notna().sum()
)

print(
    "Transactions without errors:",
    transactions["errors"].isna().sum()
)
```

```python
print(transactions["use_chip"].value_counts(dropna=False))
```
